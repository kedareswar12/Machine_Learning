# Label Encoding

## What is it?
Label encoding converts each category in a column into an **integer**. The column stays as a single column, with numbers in place of text.

| color | → | color_encoded |
|-------|---|---------------|
| red   |   | 2 |
| green |   | 1 |
| blue  |   | 0 |

Numbers are assigned in **alphabetical order** by default (blue = 0, green = 1, red = 2).

## Why use it?
- ML algorithms need numeric input, not text.
- It is simple and fast.
- It adds **no extra columns**, so memory stays low.

## The main problem
The numbers create an **artificial order**. A model may treat red (2) > green (1) > blue (0), even though colors have no ranking.

## When to use it

| Use label encoding | Avoid it |
|---|---|
| Target variable (y) in classification | Nominal features in linear models, KNN, SVM, neural networks |
| Tree-based models (Decision Tree, Random Forest, XGBoost) | Categories with no natural order |
| Categories with a natural order (Low < Medium < High) | |

## Label Encoding vs One-Hot Encoding

| | Label Encoding | One-Hot Encoding |
|---|---|---|
| Output | 1 column of integers | 1 column per category |
| Implies order? | Yes | No |
| Columns added | 0 | n (or n-1) |
| Best for | Ordinal data, target, tree models | Nominal data, linear models |

## Key points
- `LabelEncoder` is meant for the **target variable**.
- For input features, use `OrdinalEncoder`.
- For ordered categories, define the order yourself so the numbers make sense.

## One-line summary
**Label encoding = category → integer. Good for ordered data and tree models; risky for unordered data in distance-based or linear models.**

```python
import pandas as pd 
from sklearn.preprocessing import OneHotEncoder

df= pd.DataFrame(
    {
        "color" : ["red" , "blue" , 'green', "green", "red" ,"blue"]
    }
)
```

```python
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder
label_encoder = LabelEncoder()
```

```python
label_encoder.fit_transform(df[['color']])
```

```python
label_encoder.transform([['red']])
```

# Ordinal Encoding

It is used to encode categorical data that have an intrinsic order or ranking. In this technique, each category is assigned a numerical value based on its position in the order. For example, if we have a categorical variable "education level" with four possible values (high school, college, graduate, post-graduate), we can represent it using ordinal encoding as follows:

1. High school: 1
2. College: 2
3. Graduate: 3
4. Post-graduate: 4

```python
df = pd.DataFrame(
    {
        "size" : ["small" ,"medium" , "large" , "medium" , "small" , "large"]
    }
)
```

```python
Ord_encoder.fit_transform(df[['size']])
```

```python
Ord_encoder.transform([['small']])
```
