```python
# Synthetic Minority Oversampling Technique  

# IN UPSAMPLING since the data points are to be increased it will be the same points again and again 
# but in SMOTE we perform the interpolation i.e it tries to join 2 initial points and then add the random points in between them 
```

```python
"""
SMOTE (Synthetic Minority Over-sampling Technique) is an oversampling algorithm designed to resolve class imbalance in machine learning datasets. Rather than simply duplicating minority class instances (which leads to overfitting), SMOTE synthesizes new, plausible data points between existing ones
"""
```

### How It Works

Identify Minority Points: Selects a feature vector $x_i$ from the minority class at random.Find Nearest Neighbors: Calculates the $k$-nearest neighbors for $x_i$ within the same class (typically $k=5$).  Select a Neighbor: Randomly chooses one neighbor, $x_{neighbor}$.  Interpolate: Generates a new sample along the linear feature segment connecting $x_i$ and $x_{neighbor}$:  

```python
from sklearn.datasets import make_classification

X, Y = make_classification(n_samples= 1000, n_features= 2, n_redundant=0,  n_clusters_per_class= 1, weights=[0.90] , random_state=12)
```

```python
import pandas as pd

df1= pd.DataFrame(X, columns= ['f1' , 'f2'])
df2 = pd.DataFrame(Y, columns= ['target'])
final_df = pd.concat([df1, df2] , axis = 1)
final_df.head()
```

```python

oversample = SMOTE()

X ,Y = oversample.fit_resample(final_df[['f1', 'f2']], final_df['target'])
```
