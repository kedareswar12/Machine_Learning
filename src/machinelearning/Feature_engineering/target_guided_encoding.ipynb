### Target guided ordinal encoding  or Mean Target Encoding
(used to transform high-cardinality categorical variables into numerical features based on the average target value for each category)

it is a technique used to encode the categoriacal varible based ton the relationship between the target variables 



Instead of creating dozens or hundreds of binary columns (like in One-Hot Encoding) or assigning arbitrary integers (like in Standard Ordinal Encoding), Target Encoding ranks and orders categories using the expected value of the target label $y$ given the category $x = k$.

```python
import pandas as pd

df = pd.DataFrame({
    'City': ['Austin', 'Austin', 'Austin', 'Seattle', 'Seattle', 'Seattle', 'Denver', 'Austin'],
    'Bought_House': [200,150, 500, 100, 450, 320, 780, 190]
})
df
```

```python
# to get the city use the groupby command 
df.groupby('City')['Bought_House'].mean().to_dict()
```

```python
mean_price = df.groupby('City')['Bought_House'].mean().to_dict()
```

```python
df['Bought_House'] = df['City'].map(mean_price)
df['Bought_House'] 
```

```python
df[['City' , 'Bought_House']]
```
