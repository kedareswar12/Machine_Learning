```python
import numpy as np
import pandas as pd


np.random.seed(123)


n_samples = 1000
class_0_ratio = 0.9
n_class_0 = int(n_samples * class_0_ratio )
n_class_1 = n_samples - n_class_0
```

```python
n_class_0 , n_class_1
```

```python
# Create the dataframe with imbalance code 


class_0 = pd.DataFrame({
    'feature_1' : np.random.normal(loc = 0 , scale = 1 , size = n_class_0),
    'feature_2' : np.random.normal(loc = 0 , scale = 1 , size = n_class_0),
    'target' : [0] *n_class_0
})

class_1 = pd.DataFrame({
    'feature_1' : np.random.normal(loc = 0 , scale = 1 , size = n_class_1),
    'feature_2' : np.random.normal(loc = 0 , scale = 1 , size = n_class_1),
    'target' : [1] *n_class_1
})
```

```python
df = pd.concat([class_0 , class_1]).reset_index(drop=True)
df
```

```python
df['target'].value_counts()
```

```python
#upsampling

df_minority = df[df['target']==1]
df_minority
df_majority = df[df['target']==0]
```

```python
from sklearn.utils import resample 

df_minority_updated_upsample = resample (df_minority , replace = True, # sample with replacement 
            n_samples= len(df_majority),
            random_state= 42
            )
```

```python
df_minority_updated_upsample.shape
```

```python
df_upsampled  = pd.concat([df_majority , df_minority_updated_upsample])
```

```python
df_upsampled['target'].value_counts()
```
