```python
import seaborn as sns

df = sns.load_dataset('titanic')
```

```python
df.head()
```

```python
# to get the missing values 
df.isnull()
df.isnull().sum()
```

```python
# Delete the rows or data points to handle missing values 

print(df.shape
)
df.dropna().shape
```

```python
# column wise deletion 
df.dropna(axis= 1)

```

```python
# imputation missing values 

# technique -1 -> Meanvalue Imupation 
sns.histplot(df['age'] , kde= True)
```

```python
df['age'].fillna(df['age'].mean())
```

```python
df['Age_mean'] = df['age'].fillna(df['age'].mean())
```

```python
df.head
```

```python
df['Age_mean']
```

```python
df[["Age_mean" , "age"]]
```

```python
### Works well if we have the normal distribution data 
```

What if we have right skewed or ledt skewed data

```python
#### Use Median value imputation - if we have the outliers in the dataset 
```

```python
df['age_medium'] = df['age'].fillna(df['age'].median())
```

```python
# Mode value imputation- Categorial values

df[df['embarked'].isnull()]
```

```python
df.embarked.unique()
```

```python
mode_value = df[df['embarked'].notna()]['embarked'].mode()[0]
```

```python
df['embarked_mode'] = df['embarked'].fillna(mode_value) 
```

```python
df['embarked'].notna()
```
