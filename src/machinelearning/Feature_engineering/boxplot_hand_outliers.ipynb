# Five-Number Summary in Machine Learning (EDA)

The **Five-Number Summary** is a core descriptive statistics framework used during Exploratory Data Analysis (EDA). It summarizes the underlying distribution, range, and skewness of a numerical feature using five key boundary points.

---

## 1. Summary Components

| Statistic | Percentile | Definition | Primary Use in Machine Learning |
|---|---|---|---|
| **Minimum** | 0th Percentile | Lowest data point (excluding calculated outliers) | Sets baseline boundary for feature scaling/normalization |
| **First Quartile ($Q_1$)** | 25th Percentile | Boundary where 25% of data falls below it | Defines lower bound of the middle 50% of the dataset |
| **Median ($Q_2$)** | 50th Percentile | The exact midpoint of the dataset | Serves as robust central tendency metric resistant to outliers |
| **Third Quartile ($Q_3$)** | 75th Percentile | Boundary where 75% of data falls below it | Defines upper bound of the middle 50% of the dataset |
| **Maximum** | 100th Percentile | Highest data point (excluding calculated outliers) | Sets ceiling boundary for feature scaling/normalization |

---

## 2. Mathematical Applications in EDA

### Outlier Detection (Interquartile Range - IQR)
$$\text{IQR} = Q_3 - Q_1$$

Data points falling outside the following interval are flagged as statistical outliers:
$$[\text{Lower Fence}, \text{Upper Fence}] = [Q_1 - 1.5 \times \text{IQR},\; Q_3 + 1.5 \times \text{IQR}]$$

### Assessing Skewness
* **Right Skewed (Positive):** $(Q_3 - \text{Median}) > (\text{Median} - Q_1)$
* **Left Skewed (Negative):** $(Q_3 - \text{Median}) < (\text{Median} - Q_1)$
* **Symmetric:** $(Q_3 - \text{Median}) \approx (\text{Median} - Q_1)$

---

## 3. Python Implementation

```python
import pandas as pd

# Sample feature data
data = pd.Series([10, 12, 14, 15, 18, 22, 29, 31, 35, 40, 85])

# Standard 5-number summary via pandas
summary = data.describe(percentiles=[0.25, 0.5, 0.75])

print(f"Minimum:  {summary['min']}")
print(f"Q1 (25%): {summary['25%']}")
print(f"Median:   {summary['50%']}")
print(f"Q3 (75%): {summary['75%']}")
print(f"Maximum:  {summary['max']}")

````python
import numpy as np


list_marks = [45,34,5,6,3,4,52,56,77,6,75,43,23]
minimum , Q1 , median , Q3 , maximum =  np.quantile(list_marks , [0, 0.25 , 0.50 , 0.75, 1.0])
````

````python
(minimum , Q1 , median , Q3 , maximum )
````

````python
IQR = Q3 - Q1 
print(IQR)
````

````python
lower_fence = Q1 - 1.5*(IQR)
higher_fence = Q3 + 1.5*(IQR)
print(lower_fence , higher_fence)
````
